# Notebook 1: Menu Detection and OCR Extraction

```python
# Import required libraries
import os
import cv2  # Image processing
import numpy as np
from tensorflow.keras.models import load_model
from tensorflow.keras.preprocessing.image import img_to_array, load_img
from PIL import Image
import pytesseract
```

```python
# Set paths
image_folder = "C:/Users/leomo/Desktop/APP BARTINDER/Menu_Reader/data/raw_images"
model_path = "C:/Users/leomo/Desktop/APP BARTINDER/Menu_Classifier/saved_models/menu_classifier.h5"

# Load pre-trained model
menu_classifier = load_model(model_path)
```

```python
# Define image preprocessing function
def preprocess_image(image_path, target_size=(224, 224)):
    img = load_img(image_path, target_size=target_size)
    img = img_to_array(img) / 255.0  # Normalize
    img = np.expand_dims(img, axis=0)
    return img
```

```python

# Predict if image is a menu
def is_menu(image_path, threshold=0.5):
    img = preprocess_image(image_path)
    prediction = menu_classifier.predict(img)[0][0]
    return prediction >= threshold, prediction

```

```python

# Extract text using OCR
def extract_text(image_path):
    image = cv2.imread(image_path)
    gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    # Optional preprocessing
    gray_image = cv2.threshold(gray_image, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)[1]
    text = pytesseract.image_to_string(gray_image, lang='eng')
    return text
```

```python

# Run the pipeline
menu_texts = {}
for filename in os.listdir(image_folder):
    if filename.lower().endswith((".png", ".jpg", ".jpeg")):
        image_path = os.path.join(image_folder, filename)
        menu_flag, confidence = is_menu(image_path)

        print(f"Processing {filename}: {'Menu' if menu_flag else 'Not a menu'} (Confidence: {confidence:.2f})")
        
        if menu_flag:
            extracted_text = extract_text(image_path)
            menu_texts[filename] = extracted_text
            print(f"Extracted text:\n{extracted_text[:200]}...\n")
        else:
            print("Skipped text extraction.\n")

# Save extracted texts to file
import pandas as pd
df_texts = pd.DataFrame(menu_texts.items(), columns=['filename', 'extracted_text'])
df_texts.to_csv("../data/ocr_extracted_menus.csv", index=False)
```
